```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
from sklearn.datasets import load_iris
iris = load_iris()
```

```
df = pd.DataFrame()
```

```
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target

df.head()
```

```
df.shape
```

```
X = df.drop('target', axis=1)
y = df['target']
```

```
import plotly.express as px

fig = px.scatter_3d(
    df,
    x='petal length (cm)',
    y='petal width (cm)',
    z='sepal length (cm)',
    color='target',
    title="Iris Data - 3D Scatter Plot"
)

fig.show()
```

```
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import  LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

```
model1 = DecisionTreeClassifier(random_state=42)
model2 = LogisticRegression(max_iter=200, random_state=42)
model3 = KNeighborsClassifier(n_neighbors=5)
```

```
model1.fit(X, y)
model2.fit(X, y)
model3.fit(X, y)
```

```
pred1 = model1.predict(X)
pred2 = model2.predict(X)
pred3 = model3.predict(X)

# 2. Naya Table (Meta Dataframe) banana
meta_df = pd.DataFrame({
    'm1_pred': pred1,
    'm2_pred': pred2,
    'm3_pred': pred3,
    'target': y
})

# Naye table ki pehli 10 rows dekhne ke liye
print(meta_df.head(10))
```

```
from sklearn.ensemble import RandomForestClassifier

# 1. Meta-Model ke Inputs aur Target alag karna
X_meta = meta_df[['m1_pred', 'm2_pred', 'm3_pred']]
y_meta = meta_df['target']

meta_model = RandomForestClassifier(random_state=42)
meta_model.fit(X_meta, y_meta)
```

# Prediction



```
new_data = df.iloc[[0]].drop('target', axis=1)

new_data
```

```
p1 = model1.predict(new_data)[0]
p2 = model2.predict(new_data)[0]
p3 = model3.predict(new_data)[0]

print(f"Decision Tree Prediction (m1) : {p1}")
print(f"Logistic Reg Prediction  (m2) : {p2}")
print(f"KNN Prediction           (m3) : {p3}")
```

```
meta_input = pd.DataFrame({
    'm1_pred': [p1],
    'm2_pred': [p2],
    'm3_pred': [p3]
})

final_pred = meta_model.predict(meta_input)[0]

print(f"Meta-Model Final Prediction: {final_pred}")
```
